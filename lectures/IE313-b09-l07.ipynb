### Training/Testing Set

Tách bộ dữ liệu thành:

- Training set (70%).
- Testing set (30%).
- Phát triển mô hình với tập training.
- Dùng testing để đánh giá hiệu suất dự đoán của mô hình.
- Khi hoàn thành đánh giá nên sử dụng alị tất cả dữ liệu để phát triển lại mô hình để có hiệu suất tốt nhất.

#### Hàm `train_test_split()`

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x_data,
    y_data,
    test_size = 0.3,
    random_state = 0
)
```

- `x_data`: biến độc lập
- `y_data`: biến phụ thuộc (`df['price']`)
- `x_train`, `y_train`: biến độc lập và phụ thuộc của bộ `train`
- `x_test`, `y_test`: biến độc lập và phụ thuộc của bộ `test`
- `test_size`: kích cỡ của bộ `test`, đơn vị % (`0.3`, `0.4`, vv..)
- `random_state`: số dùng để làm cơ sở kiểm soát việc xáo trộn dữ liệu (seed?)

## 2 - Overfitting - Underfitting - Chọn Mô Hình

- Overfitting
- Underfitting
- Chọn mô hình
- Minh họa

### Overfitting

- Có kết quả tốt cho bộ train nhưng không tốt cho bộ test.

Nguyên nhân:

- Dữ liệu nhiễu, hoặc bất thường (outliner) trong train. $\Rightarrow$ làm mềm đường hồi quy.

### Underfitting

- Có kết quả không tốt cho bộ test sau khi phát triển.

Nguyên nhân:

- Tập dữ liệu chưa đạt chuẩn.

Khắc phục:

- Phân tích thăm dò lại bộ dữ liệu.
- Bổ sung thêm dữ liệu train.
- Đổi thuật toán.

### Good Fitting

- Vừa khớp.
- Nằm giữa Overfitting và Undefitting.

### Ví Dụ

````python
orders = range(1, 12)

x5 = df_theory[['horsepower']]
y5 = df_theory['price']

r2_train = []
r2_test = []

for order in orders:
    x5_train, x5_test, y5_train, y5_test = train_test_split(
        x5,
        y5,
        test_size = 0.2,
        random_state = 0
    )

    m5 = LinearRegression()
    pf5 = PolynomialFeatures(degree=order, include_bias=False)

    x5_pf_train = pf5.fit_transform(x5_train)
    x5_pf_test = pf5.fit_transform(x5_test)

    m5.fit(x5_pf_train, y5_train)

    r2_train.append(m5.score(x5_pf_train, y5_train))
    r2_test.append(m5.score(x5_pf_test, y5_test))
````

````python
r2_train
````

````python
r2_test
````

````python
plt.title("r2_train")
plt.plot(orders, r2_train, label='train')
plt.legend(
    loc='upper left',
    bbox_to_anchor=(1, 1),
    fancybox=True,
    # shadow=True
)
````

````python
plt.title("r2_test")
plt.plot(orders, r2_test, label='test', color="orange")
plt.legend(
    loc='upper left',
    bbox_to_anchor=(1, 1),
    fancybox=True,
    # shadow=True
)
````

````python
plt.title("So sánh r2_train vs r2_test")
plt.plot(orders, r2_train, label='train')
plt.plot(orders, r2_test, label='test')
plt.grid(which='both')
plt.legend(
    loc='upper left',
    bbox_to_anchor=(1, 1),
    fancybox=True,
    # shadow=True
)
````

Nhận Xét

Càng tăng bậc:

- mô hình càng "overfitting" tập train, nhưng có giảm ở đoạn từ 5-7.
- mô hình càng thiếu phù hợp với tập test, đặc biệt từ bậc 3-5.
